```python
import pandas as pd
import xgboost as xgb
from sklearn.preprocessing import LabelEncoder

# Load dataset for encoders
df = pd.read_csv("data_bodmas.csv")

# Prepare label encoders using original data
categorical_cols = ['Donor_BloodType', 'Recipient_BloodType', 'Donor_Rh', 'Recipient_Rh', 'Organ_Type']
label_encoders = {}

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le

# Re-train model for this session
X = df.drop(columns=['Donor_ID', 'Recipient_ID', 'Is_Match'])
y = df['Is_Match']
model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
model.fit(X, y)

import pickle

# Load the trained model
model = pickle.load(open("organ_match_model.pkl", "rb"))

# Load the label encoders
label_encoders = pickle.load(open("label_encoders.pkl", "rb"))

# Prediction function
def predict_match(new_data: dict):
    # Convert to DataFrame
    input_df = pd.DataFrame([new_data])

    # Encode categorical features using original encoders
    for col in categorical_cols:
        le = label_encoders[col]
        input_df[col] = le.transform([new_data[col]])  # ensure consistent encoding

    # Predict
    prediction = model.predict(input_df)[0]
    return "Match" if prediction == 1 else "No Match"

# Interactive input function
def get_user_input():
    print("Please enter the following details:")

    donor_blood_type = input("Donor Blood Type (e.g. A, B, AB, O): ")
    recipient_blood_type = input("Recipient Blood Type (e.g. A, B, AB, O): ")
    donor_rh = input("Donor Rh (e.g. + or -): ")
    recipient_rh = input("Recipient Rh (e.g. + or -): ")
    organ_type = input("Organ Type (e.g. Kidney, Heart, etc.): ")
    donor_age = int(input("Donor Age: "))
    recipient_age = int(input("Recipient Age: "))
    donor_lat = float(input("Donor Latitude (e.g. 25.5): "))
    donor_lon = float(input("Donor Longitude (e.g. 80.1): "))
    recipient_lat = float(input("Recipient Latitude (e.g. 25.6): "))
    recipient_lon = float(input("Recipient Longitude (e.g. 80.0): "))

    # Prepare input dictionary for prediction
    new_data = {
        "Donor_BloodType": donor_blood_type,
        "Recipient_BloodType": recipient_blood_type,
        "Donor_Rh": donor_rh,
        "Recipient_Rh": recipient_rh,
        "Organ_Type": organ_type,
        "Donor_Age": donor_age,
        "Recipient_Age": recipient_age,
        "Donor_Lat": donor_lat,
        "Donor_Lon": donor_lon,
        "Recipient_Lat": recipient_lat,
        "Recipient_Lon": recipient_lon
    }

    return new_data

# Main flow
new_donor_recipient = get_user_input()
result = predict_match(new_donor_recipient)
print("Prediction Result:", result)




```
